```python
from huggingface_hub import list_repo_files

repo_id = "SpatialVID/SpatialVID"
files = list_repo_files(repo_id, repo_type="dataset")

# Find candidate metadata CSVs
csvs = [f for f in files if f.lower().endswith(".csv")]
print("CSV files found:", len(csvs))
for f in csvs[:50]:
    print(" -", f)

# Find likely metadata file(s)
candidates = [f for f in csvs if "metadata" in f.lower()]
print("\nMetadata candidates:")
for f in candidates:
    print(" -", f)
```

```python
from huggingface_hub import hf_hub_download
import pandas as pd

# Prefer the known path if present, else fallback to first candidate
preferred = "data/train/SpatialVID_metadata.csv"
meta_path_in_repo = preferred if preferred in files else (candidates[0] if candidates else csvs[0])

meta_local = hf_hub_download(
    repo_id=repo_id,
    repo_type="dataset",
    filename=meta_path_in_repo
)
print("Downloaded metadata:", meta_local)

df = pd.read_csv(meta_local)
print("Rows:", len(df))
print("Columns:", df.columns.tolist()[:40])
df.head()
```

```python
# Keep rows whose sceneType looks like nature/landscape
scene = df["sceneType"].fillna("").str.lower()

nature_mask = scene.str.contains("natural") | scene.str.contains("landscape") | scene.str.contains("waterfront") | scene.str.contains("hillside") | scene.str.contains("forest") | scene.str.contains("mountain") | scene.str.contains("beach") | scene.str.contains("river") | scene.str.contains("lake")

df_nat = df[nature_mask].copy()

# Optional: prefer higher aesthetic score (if present)
if "aesthetic score" in df_nat.columns:
    df_nat = df_nat.sort_values("aesthetic score", ascending=False)

print("Nature-ish rows:", len(df_nat))
df_nat[["id","group id","video path","sceneType","weather","timeOfDay"]].head(10)
```

```python
# Pick one row and get its group id
row = df_nat.iloc[0]
gid = int(row["group id"])
gid
```

```python
from huggingface_hub import hf_hub_download

repo_id = "SpatialVID/SpatialVID"
group_name = f"group_{gid:04d}"
tar_in_repo = f"videos/{group_name}.tar.gz"

tar_local = hf_hub_download(
    repo_id=repo_id,
    repo_type="dataset",
    filename=tar_in_repo
)
print("Downloaded:", tar_local)
```

```python
import os, tarfile

extract_root = f"/content/{group_name}"
os.makedirs(extract_root, exist_ok=True)

with tarfile.open(tar_local, "r:gz") as tar:
    tar.extractall(path=extract_root)

print("Extracted to:", extract_root)
```

```python
import glob
from IPython.display import Video, display

# Find mp4s anywhere under the extracted folder
mp4s = sorted(glob.glob(f"{extract_root}/**/*.mp4", recursive=True))
print("MP4s found:", len(mp4s))

for p in mp4s[:5]:
    print("Preview:", p)
    display(Video(p, embed=True))
```
